```python
import os
import shutil
import logging
```

```python
def create_manga_volumes(working_directory, manga_name):
    logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
    volumes_folder = os.path.join(working_directory, f"{manga_name} Volumes")

    # Check if volumes folder already exists
    if os.path.exists(volumes_folder):
        logging.info(f"'{volumes_folder}' already exists. Skipping volume creation.")
        return

    manga_path = os.path.join(working_directory, manga_name)
    chapters = [folder for folder in os.listdir(manga_path) if os.path.isdir(os.path.join(manga_path, folder))]

    volumes = {}

    # Organize chapters by volume
    logging.info("Organizing chapters by volume")
    for chapter in chapters:
        if "Vol." in chapter and "Ch." in chapter:
            volume = chapter.split("Vol.")[1].split()[0]
            chapter_num = chapter.split("Ch.")[1].split()[0]
            try:
                volume_int = int(volume)
                chapter_int = int(chapter_num)
                if volume_int not in volumes:
                    volumes[volume_int] = []
                volumes[volume_int].append((chapter_int, chapter))
            except ValueError:
                logging.warning(f"Skipping chapter with invalid volume or chapter number: {chapter}")
                continue

    # Process each volume
    os.makedirs(volumes_folder, exist_ok=True)

    for volume, chapter_list in volumes.items():
        logging.info(f"Processing volume {volume}")
        chapter_list.sort(key=lambda x: x[0])
        volume_folder = os.path.join(volumes_folder, f"Vol.{volume:02d}")
        os.makedirs(volume_folder, exist_ok=True)

        page_counter = 1
        for _, chapter in chapter_list:
            logging.info(f"Processing chapter {chapter}")
            chapter_path = os.path.join(manga_path, chapter)
            pages = sorted([f for f in os.listdir(chapter_path) if f.endswith('.jpg')])

            for page in pages:
                src = os.path.join(chapter_path, page)
                dst_filename = f"{page_counter:04d}.jpg"
                dst = os.path.join(volume_folder, dst_filename)
                shutil.copy(src, dst)
                logging.info(f"Copied {src} to {dst}")
                page_counter += 1

    logging.info("Manga volumes created successfully.")

def save_selected_pages(working_directory, manga_name, volume_number, pages_to_save):
    # Ensure pages_to_save is sorted and contains unique values
    if pages_to_save != sorted(set(pages_to_save)):
        raise ValueError("The pages_to_save list must be sorted and contain no duplicates.")

    logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
    source_folder = os.path.join(working_directory, f"{manga_name} Volumes", f"Vol.{volume_number:02d}")
    output_folder = os.path.join(r"C:\Hobby\Editing\Output\Manga", manga_name, f"Vol.{volume_number:02d}")
    os.makedirs(output_folder, exist_ok=True)

    for page in pages_to_save:
        found = False
        for ext in ['.jpg', '.jpeg']:
            src_file = os.path.join(source_folder, f"{page:04d}{ext}")
            dst_file = os.path.join(output_folder, f"{page:04d}{ext}")
            if os.path.exists(src_file):
                shutil.copy(src_file, dst_file)
                logging.info(f"Copied {page:04d}{ext} of volume {volume_number}")
                found = True
                break
        
        if not found:
            logging.warning(f"Page {page:04d}.jpg or .jpeg does not exist in {source_folder}")

    logging.info("Selected pages saved successfully.")
```
